```python
import os
import glob
import shutil

EXTRACTED_PAPERS_DIR = "/media/gpu1-3e82-0/ssd/Github Folder/read_papers/extracted_papers"
UPLOAD_DIR = "/media/gpu1-3e82-0/ssd/Github Folder/read_papers/upload_folder"

def copy_for_upload(base_dir=EXTRACTED_PAPERS_DIR, upload_dir=UPLOAD_DIR, paper_folder=None):
    """
    Copy everything EXCEPT original.jpg and result_with_boxes.jpg from each page_n folder,
    plus full_extracted.md from the paper root, into upload_folder preserving directory structure.

    Args:
        base_dir: Root extracted_papers directory.
        upload_dir: Destination directory mirroring the same structure.
        paper_folder: If provided, only process that specific paper folder (name, not full path).
                      If None, process all paper folders.
    """
    skip = {"original.jpg", "result_with_boxes.jpg"}
    copied = 0

    if paper_folder:
        paper_dirs = [os.path.join(base_dir, paper_folder)]
    else:
        paper_dirs = sorted(glob.glob(os.path.join(base_dir, "*")))

    for paper_dir in paper_dirs:
        if not os.path.isdir(paper_dir):
            continue
        paper_name = os.path.basename(paper_dir)

        # Copy full_extracted.md
        md_src = os.path.join(paper_dir, "full_extracted.md")
        if os.path.isfile(md_src):
            md_dst_dir = os.path.join(upload_dir, paper_name)
            os.makedirs(md_dst_dir, exist_ok=True)
            shutil.copy2(md_src, os.path.join(md_dst_dir, "full_extracted.md"))
            copied += 1

        # Copy page contents (excluding original.jpg and result_with_boxes.jpg)
        pages_dir = os.path.join(paper_dir, "pages")
        if not os.path.isdir(pages_dir):
            continue
        for page in os.listdir(pages_dir):
            page_path = os.path.join(pages_dir, page)
            if not os.path.isdir(page_path) or not page.startswith("page_"):
                continue
            for entry in os.listdir(page_path):
                if entry in skip:
                    continue
                src = os.path.join(page_path, entry)
                dst_dir = os.path.join(upload_dir, paper_name, "pages", page)
                os.makedirs(dst_dir, exist_ok=True)
                dst = os.path.join(dst_dir, entry)
                if os.path.isdir(src):
                    shutil.copytree(src, dst, dirs_exist_ok=True)
                else:
                    shutil.copy2(src, dst)
                copied += 1

    print(f"Copied {copied} items to {upload_dir}.")
```

```python
copy_for_upload(paper_folder="3D_Gaussian_Flats_Hybrid_2D3D_Photometric_Scene_Reconstruction_uVxQEIgXfL")
```
